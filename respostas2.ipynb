#### 4)  percentual de representação que cada estado teve

```python
faturamento_estados = {
    "SP": 67836.43,
    "RJ": 36678.66,
    "MG": 29229.88,
    "ES": 27165.48,
    "Outros": 19849.53
}

# Calcula o valor total do faturamento
faturamento_total = sum(faturamento_estados.values())

# Calcula o percentual de representação de cada estado
percentuais = {estado: (valor / faturamento_total) * 100 for estado, valor in faturamento_estados.items()}

# Exibir resultados
for estado, percentual in percentuais.items():
    print(f"{estado}: {percentual:.2f}%")
```

##### 5) Program que inverte string

```python
# Frase da canção 'Bob' de Wierd Al, é um polindrome. Lê-se igual de tras para frente
frase = "God! A red nugget! A fat egg under a dog!"


def inverter_string(s):
    invertida = ""
    # Percorre a string de trás para frente e monta a nova string
    for i in range(len(s) - 1, -1, -1):
        invertida += s[i]
    return invertida


frase_invertida = inverter_string(frase)


print(f"Frase original: {frase}")
print(f"Frase invertida: {frase_invertida}")
```
